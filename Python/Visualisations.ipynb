```python
from scipy.stats import percentileofscore

metrics = [
    'Goals_P90',
    'xG_P90',
    'Goal_Conversion',
    'Assists_P90',
    'xA_P90',
    'KP_P90'
]

def get_percentiles(player_name):

    row = strikers[
        strikers['player_name'] == player_name
    ]

    vals = []

    for metric in metrics:

        pct = percentileofscore(
            strikers[metric],
            row[metric].iloc[0]
        )

        vals.append(round(pct,1))

    return vals
```

```python
import os
import pandas as pd

file_path = os.path.join(path, "players_data-2025_2026.csv")

new_data = pd.read_csv(file_path)

print("Rows:", len(new_data))
print("Columns:", len(new_data.columns))

print("\nCompetitions available:")
print(new_data["Comp"].value_counts())
```

```python
# Clean league names
league_name_map = {
    "eng Premier League": "Premier League",
    "es La Liga": "La Liga",
    "de Bundesliga": "Bundesliga",
    "it Serie A": "Serie A",
    "fr Ligue 1": "Ligue 1"
}

new_data["League"] = new_data["Comp"].map(league_name_map)

print(new_data["League"].value_counts())
```

```python
print(new_data.columns.tolist())
```

```python
# Keep only the new columns we need
defense_data = new_data[
    [
        "Player",
        "Squad",
        "League",
        "Age",
        "Int",
        "TklW"
    ]
].copy()

# Rename columns to match our project style
defense_data.rename(columns={
    "Player": "player_name",
    "Squad": "team_title",
    "Age": "age",
    "Int": "interceptions",
    "TklW": "tackles_won"
}, inplace=True)

print(defense_data.shape)
defense_data.head()
```

```python
import re
import unicodedata

def clean_player_name(name):
    if pd.isna(name):
        return ""

    name = str(name).lower().strip()

    # Remove accents: João -> joao
    name = unicodedata.normalize("NFKD", name)
    name = "".join(c for c in name if not unicodedata.combining(c))

    # Keep only letters and numbers
    name = re.sub(r"[^a-z0-9 ]", "", name)

    # Remove extra spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name


df_all["match_name"] = df_all["player_name"].apply(clean_player_name)
defense_data["match_name"] = defense_data["player_name"].apply(clean_player_name)

print("df_all rows:", len(df_all))
print("Defense rows:", len(defense_data))

print("\nExample df_all names:")
print(df_all[["player_name", "match_name"]].head())

print("\nExample defense names:")
print(defense_data[["player_name", "match_name"]].head())
```

```python
# Exact name matching check
defense_names = set(defense_data["match_name"])

df_all["exact_match"] = df_all["match_name"].isin(defense_names)

print("Exact matches:", df_all["exact_match"].sum())
print("Unmatched:", (~df_all["exact_match"]).sum())
print("Match percentage:", round(df_all["exact_match"].mean() * 100, 2), "%")
```

```python
df_all.loc[
    ~df_all["exact_match"],
    ["player_name", "League", "team_title"]
].head(30)
```

```python
# Reset fuzzy matching
df_all["final_match_name"] = df_all["match_name"]
df_all["match_score"] = 100.0

# Make team names clean in both datasets
df_all["team_match"] = (
    df_all["team_title"]
    .astype(str)
    .str.lower()
    .str.strip()
)

defense_data["team_match"] = (
    defense_data["team_title"]
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
from rapidfuzz import process, fuzz

fixed = 0

for idx, row in df_all.loc[~df_all["exact_match"]].iterrows():

    league = row["League"]
    team = row["team_match"]
    name = row["match_name"]

    # Candidates from SAME LEAGUE + SAME TEAM only
    candidates = defense_data.loc[
        (defense_data["League"] == league) &
        (defense_data["team_match"] == team),
        "match_name"
    ].dropna().unique().tolist()

    if not candidates:
        continue

    result = process.extractOne(
        name,
        candidates,
        scorer=fuzz.WRatio,
        score_cutoff=80
    )

    if result is not None:
        matched_name, score, _ = result

        df_all.at[idx, "final_match_name"] = matched_name
        df_all.at[idx, "match_score"] = float(score)

        fixed += 1

print("Safe fuzzy matches found:", fixed)
```

```python
# Check how many of the 2775 df_all players have a safe match

df_all["ready_to_merge"] = (
    df_all["exact_match"] |
    (df_all["match_score"] >= 90)
)

print("Total players in df_all:", len(df_all))
print("Matched df_all players:", df_all["ready_to_merge"].sum())
print("Unmatched df_all players:", (~df_all["ready_to_merge"]).sum())
```

```python
# Create the final safe matching name

df_all["safe_match_name"] = df_all["match_name"]

# Replace with fuzzy-matched name ONLY for strong matches
mask = (
    (~df_all["exact_match"]) &
    (df_all["match_score"] >= 90)
)

df_all.loc[mask, "safe_match_name"] = df_all.loc[
    mask, "final_match_name"
]
```

```python
# Create unique merge key using player + team + league

df_all["merge_key"] = (
    df_all["safe_match_name"].astype(str) + "|" +
    df_all["team_match"].astype(str) + "|" +
    df_all["League"].astype(str)
)

defense_data["merge_key"] = (
    defense_data["match_name"].astype(str) + "|" +
    defense_data["team_match"].astype(str) + "|" +
    defense_data["League"].astype(str)
)

# Keep only required columns
defense_merge = defense_data[
    [
        "merge_key",
        "age",
        "interceptions",
        "tackles_won"
    ]
].copy()

# Check duplicate full keys
print("Duplicate merge keys:", defense_merge["merge_key"].duplicated().sum())
```

```python
# Create final key: matched name + league + team

df_all["merge_key3"] = (
    df_all["safe_match_name"].astype(str).str.strip()
    + " | " +
    df_all["League"].astype(str).str.strip()
    + " | " +
    df_all["team_title"].astype(str).str.strip()
)

defense_data["merge_key3"] = (
    defense_data["match_name"].astype(str).str.strip()
    + " | " +
    defense_data["League"].astype(str).str.strip()
    + " | " +
    defense_data["team_title"].astype(str).str.strip()
)

print("df_all rows:", len(df_all))
print("df_all unique keys:", df_all["merge_key3"].nunique())

print("\nDefense rows:", len(defense_data))
print("Defense unique keys:", defense_data["merge_key3"].nunique())

print("\nDuplicate defense keys:",
      defense_data["merge_key3"].duplicated().sum())
```

```python
# Remove the defensive columns from the previous incorrect merge
cols_to_remove = [
    "age",
    "interceptions",
    "tackles_won"
]

df_all.drop(
    columns=[col for col in cols_to_remove if col in df_all.columns],
    inplace=True
)

# Prepare the correct defense table
defense_final = defense_data[
    [
        "merge_key3",
        "age",
        "interceptions",
        "tackles_won"
    ]
].copy()

# Correct merge using NAME + LEAGUE + TEAM
df_all = df_all.merge(
    defense_final,
    on="merge_key3",
    how="left",
    validate="many_to_one"
)

# Check result
print("Rows after merge:", len(df_all))
print("Players with defensive data:",
      df_all["interceptions"].notna().sum())
print("Players without defensive data:",
      df_all["interceptions"].isna().sum())

df_all[
    [
        "player_name",
        "team_title",
        "League",
        "age",
        "interceptions",
        "tackles_won"
    ]
].head(10)
```

```python
# ==============================
# 1. FIX MISSING AGE VALUES
# ==============================

# Create one age record per player + league
age_lookup = (
    defense_data[
        ["match_name", "League", "age"]
    ]
    .dropna(subset=["age"])
    .drop_duplicates(
        subset=["match_name", "League"]
    )
    .rename(columns={
        "match_name": "safe_match_name",
        "age": "age_fixed"
    })
)

# Merge age lookup
df_all = df_all.merge(
    age_lookup,
    on=["safe_match_name", "League"],
    how="left"
)

# Fill only missing age values
df_all["age"] = df_all["age"].fillna(
    df_all["age_fixed"]
)

# Remove helper age column
df_all.drop(
    columns=["age_fixed"],
    inplace=True
)


# ==============================
# 2. CHECK FINAL DATA
# ==============================

print("Total rows:", len(df_all))
print("Age available:", df_all["age"].notna().sum())
print("Age missing:", df_all["age"].isna().sum())

print("\nDefensive data available:",
      df_all["interceptions"].notna().sum())

print("Defensive data missing:",
      df_all["interceptions"].isna().sum())


# ==============================
# 3. CREATE POWER BI EXCEL FILE
# ==============================

powerbi_defense = df_all[
    [
        "player_name",
        "team_title",
        "League",
        "age",
        "interceptions",
        "tackles_won"
    ]
].copy()

# Export
file_name = "Football_Defense_2025_26.xlsx"

powerbi_defense.to_excel(
    file_name,
    index=False
)

print("\nExcel file created:", file_name)
print("Rows exported:", len(powerbi_defense))
```
